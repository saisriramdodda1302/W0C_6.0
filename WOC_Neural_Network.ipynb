```python
# Imported necessary libraries
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt

# Read data from a CSV file
data = pd.read_csv('/content/drive/MyDrive/Classification_train.csv')

# Converted data to a numpy array
data = np.array(data)

# Got the dimensions of the data
m, n = data.shape #m=30000...n=784+1

# Shuffled the data before splitting into development and training sets
np.random.shuffle(data) #The data is shuffled randomly to introduce variability and avoid any order-related bias.

# Extracted the first 1000 rows for development data
data_for_development = data[0:1000].T   #here data_for_development.shape =(785, 1000)
Y_for_development = data_for_development[0]
X_for_development = data_for_development[1:n]
X_for_development = X_for_development / 255.

# Extracted the remaining rows for training data
data_train = data[1000:m].T
Y_train = data_train[0]
X_train = data_train[1:n]
X_train = X_train / 255.

# Defined the number of classes and the input size for the neural network
_, m_train = X_train.shape  #nothing but representing the m_train value =29000

# Initialized parameters for the neural network
def init_params():
    W1 = np.random.rand(10, 784) - 0.5  #subtracting 0.5 because we want to center values around 0..
    b1 = np.random.rand(10, 1) - 0.5    #as elements values have value from 0-1.excluding1 and including 0
    W2 = np.random.rand(10, 10) - 0.5
    b2 = np.random.rand(10, 1) - 0.5  #all have 10 rows
    return W1, b1, W2, b2

# Defined the ReLU activation function
def ReLU(Z):
    return np.maximum(Z, 0)

# Defined the softmax activation function
def softmax(Z):
    A = np.exp(Z) / sum(np.exp(Z))
    return A

# Implemented the forward propagation step of the neural network
def forward_prop(W1, b1, W2, b2, X):
    Z1 = W1.dot(X) + b1
    A1 = ReLU(Z1)
    Z2 = W2.dot(A1) + b2
    A2 = softmax(Z2)
    return Z1, A1, Z2, A2

# Defined the derivative of the ReLU activation function
def ReLU_deriv(Z):
    return Z > 0

# One-hot encoded the target variable Y
def one_hot(Y):
    one_hot_Y = np.zeros((Y.size, Y.max() + 1))
    one_hot_Y[np.arange(Y.size), Y] = 1  #it means go to each row ,specify the column of Y and indicate it as 1
    one_hot_Y = one_hot_Y.T
    return one_hot_Y

# Implemented the backward propagation step of the neural network
def backward_prop(Z1, A1, Z2, A2, W1, W2, X, Y):
    one_hot_Y = one_hot(Y)
    dZ2 = A2 - one_hot_Y #Computing the error in the output layer by subtracting the one-hot encoded matrix from the actual output A2.
    dW2 = 1 / m * dZ2.dot(A1.T)
    db2 = 1 / m * np.sum(dZ2)
    dZ1 = W2.T.dot(dZ2) * ReLU_deriv(Z1)
    dW1 = 1 / m * dZ1.dot(X.T)
    db1 = 1 / m * np.sum(dZ1)
    return dW1, db1, dW2, db2

# Updated the parameters using gradient descent
def update_params(W1, b1, W2, b2, dW1, db1, dW2, db2, alpha):
    W1 = W1 - alpha * dW1
    b1 = b1 - alpha * db1
    W2 = W2 - alpha * dW2
    b2 = b2 - alpha * db2
    return W1, b1, W2, b2

# Got predictions from the output layer
def get_predictions(A2): #np.argmax(A2, 0) is the NumPy function that identifies the index of the maximum value along the specified axis.
    return np.argmax(A2, 0)

# Calculated accuracy based on predictions and actual labels
def get_accuracy(predictions, Y):
    print(predictions, Y)
    return np.sum(predictions == Y) / Y.size

# Implemented gradient descent to train the neural network
def gradient_descent(X, Y, alpha, iterations):
    W1, b1, W2, b2 = init_params()
    for i in range(iterations):
        Z1, A1, Z2, A2 = forward_prop(W1, b1, W2, b2, X)
        dW1, db1, dW2, db2 = backward_prop(Z1, A1, Z2, A2, W1, W2, X, Y)
        W1, b1, W2, b2 = update_params(W1, b1, W2, b2, dW1, db1, dW2, db2, alpha)
        if i % 10 == 0:
            print("Iterations: ", i)
            predictions = get_predictions(A2)
            print(get_accuracy(predictions, Y))
    return W1, b1, W2, b2

# Trained the neural network using gradient descent
W1, b1, W2, b2 = gradient_descent(X_train, Y_train, 0.10, 500)

# Made predictions on specific examples from the training set
def make_predictions(X, W1, b1, W2, b2):
    _, _, _, A2 = forward_prop(W1, b1, W2, b2, X)
    predictions = get_predictions(A2) #The fourth value returned by forward_prop is assigned to the variable A2
    return predictions

# Tested predictions on a few examples from the training set
def test_prediction(index, W1, b1, W2, b2):
    current_image = X_train[:, index, None]
    prediction = make_predictions(X_train[:, index, None], W1, b1, W2, b2)
    label = Y_train[index]
    print("Prediction: ", prediction)
    print("Label: ", label)

    current_image = current_image.reshape((28, 28)) * 255
    plt.gray()
    plt.imshow(current_image, interpolation='nearest')
    plt.show()

# Tested predictions on specific examples from the training set
test_prediction(0, W1, b1, W2, b2)
test_prediction(1, W1, b1, W2, b2)
test_prediction(2, W1, b1, W2, b2)
test_prediction(3, W1, b1, W2, b2)

# Made predictions on the development set and calculated accuracy
for_development_predictions = make_predictions(X_for_development, W1, b1, W2, b2)
get_accuracy(for_development_predictions, Y_for_development)
```

```python
# Loaded and preprocessed the test data
test_data = pd.read_csv('/content/drive/MyDrive/Classification_test.csv')
test_data = np.array(test_data)
data_test = test_data.T
X_test = data_test[1:]
X_test = X_test / 255.

# Made predictions on the test dataset
test_predictions = make_predictions(X_test, W1, b1, W2, b2)

# Displayed predictions for all test samples
all_predictions = test_predictions.tolist()
print("All Predictions:", all_predictions)




```
