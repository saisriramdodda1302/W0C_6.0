```python
from google.colab import drive
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path_train='/content/drive/MyDrive/Lineardata_train.csv'
path_test='/content/drive/MyDrive/Lineardata_test.csv'
```

```python
train=pd.read_csv(path_train)
test=pd.read_csv(path_test)
train.head(50000)
```

```python
test.head(10000)
```

```python
train_data=train.values
Y=train_data[:,0].reshape(train_data.shape[0],1) #here Y is the 1st column of the given train set(target)
#1: This specifies that the new array will have only one column.
X=train_data[:,1:] #it is an array containing all 20 feautures and 50000 samples
```

```python
test_data=test.values
Y_test=test_data[:,0].reshape(test_data.shape[0],1) #here Y is the 1st column of the given test set(target)
X_test=test_data[:,1:] #it is an array containing all 20 feautures and 10000 samples
```

```python
X=np.vstack((np.ones((X.shape[0],)),X.T)).T #it makes to add a column of 1's infront of 20 features,
#np.vstack is a function in the NumPy library used to combine multiple arrays vertically.
X_test=np.vstack((np.ones((X_test.shape[0],)),X_test.T)).T #because it helps in finding intercept
```

```python
print("shape of X_train:",X.shape)
print("shape of Y_train:",Y.shape)
print("shape of X_test:",X_test.shape)
print("shape of Y_test:",Y_test.shape)
```

```python
def model(X,Y,learning_rate,iteration):
  m=Y.size
  w=np.zeros((X.shape[1],1))  # an array having one column and 21 rows of zeros
  cost_list=[]


  for i in range (iteration):

    y_pred = np.dot(X,w)
    cost=(1/(2*m))*np.sum(np.square(y_pred  - Y))

    d_w=(1/m)*np.dot(X.T,y_pred - Y) # we used X.T beacuse of matrix multiplication
    w = w-learning_rate*d_w

    cost_list.append(cost)

    if i % (iteration // 10) == 0:  #as i iterations can be divided into 10 parts
      print("cost after ",i,"iterations is",cost)
  return w,cost_list
  print(w)
```

```python
iteration = 50000  # Increase the number of iterations
learning_rate = 0.000001    # Adjust the learning rate

w,cost_list = model(X,Y,learning_rate=learning_rate, iteration=iteration)
#cost_list=model(X,Y,learning_rate=learning_rate, iteration=iteration)
```

```python
y_pred = np.dot(X,w)
y_pred_test=np.dot(X_test,w)  #X.shape[0] =50000
mean_absolute_error = (1 / X.shape[0]) * np.sum(np.abs(y_pred - Y))
```

```python
# Assuming you have actual values (Y_true) and predicted values (Y_pred)
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Calculating metrics
mse = mean_squared_error(Y, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(Y, y_pred)
r2 = r2_score(Y, y_pred)

# Printing or visualizing the results

print("Mean Absolute Error:", mae)
print("R-squared Score:", r2)
```
