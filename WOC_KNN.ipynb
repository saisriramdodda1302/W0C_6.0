```python
import numpy as np
import pandas as pd

class KNN:
    def __init__(self, k=4):
        self.k = k

    def fit(self, X, y):
        self.X_train = X
        self.y_train = y

    def predict(self, X):
        predictions = [self._predict(x) for x in X]
        return predictions

    def _predict(self, x):
        # Compute the distance using vectorized operations
        distances = np.sqrt(np.sum((self.X_train - x)**2, axis=1))

        # Get the closest k
        k_indices = np.argsort(distances)[:self.k]  #np.argsort-Finds the indices of the k-nearest neighbors by sorting the distances.
        k_nearest_labels = self.y_train[k_indices] #k_nearest_labels is an array representing the labels of the k-nearest neighbors

        # Majority vote without using Counter
        unique_labels, counts = np.unique(k_nearest_labels, return_counts=True)
        most_common_label = unique_labels[np.argmax(counts)]#the most common label among the neighbors using np.argmax
        return most_common_label

# Load the CSV data
data = pd.read_csv("/content/drive/MyDrive/Classification_train.csv")

# Separate features and labels (considering the first column as the label)
X = data.iloc[:, 1:].values# Extracted features from the DataFrame, excluding the first column
y = data.iloc[:, 0].values#1st column _labels

# Create a KNN classifier
knn = KNN(k=4)

# Train the classifier
knn.fit(X, y)

# Extract features, skipping the first column (ID)
new_data = pd.read_csv("/content/drive/MyDrive/Classification_test.csv")
new_features = new_data.iloc[:, 1:].values

# Make predictions
predictions = knn.predict(new_features)

# Print the predictions
print(predictions)
#24min
```

```python
import numpy as np
import pandas as pd

class KNN:
    def __init__(self, k=4):
        self.k = k

    def fit(self, X, y):
        self.X_train = X
        self.y_train = y

    def predict(self, X):
        predictions = [self._predict(x) for x in X]
        return predictions

    def _predict(self, x):
        # Compute the distance using vectorized operations
        distances = np.sqrt(np.sum((self.X_train - x)**2, axis=1))

        # Get the closest k
        k_indices = np.argsort(distances)[:self.k]
        k_nearest_labels = self.y_train[k_indices]

        # Majority vote without using Counter
        unique_labels, counts = np.unique(k_nearest_labels, return_counts=True)
        most_common_label = unique_labels[np.argmax(counts)]
        return most_common_label

# Load the training data
data = pd.read_csv("/content/drive/MyDrive/Classification_train.csv")  # Load train data only

# Split into training and test sets
train_data = data.iloc[:15000, :]  # First 15000 rows as training data
test_data = data.iloc[15000:, :]   # Remaining rows as test data

# Separate features and labels
X_train = train_data.iloc[:, 1:].values
y_train = train_data.iloc[:, 0].values
X_test = test_data.iloc[:, 1:].values
y_test = test_data.iloc[:, 0].values  # Labels for test data to calculate accuracy

# Create a KNN classifier
knn = KNN(k=4)

# Train the classifier
knn.fit(X_train, y_train)

# Make predictions on the test set
predictions = knn.predict(X_test)

# Calculate and print accuracy
accuracy = np.mean(predictions == y_test)
print("Accuracy:", accuracy)

#18min
```
